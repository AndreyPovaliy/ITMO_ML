# Диаграмма последовательности (Sequence Diagram)

![](./img/uml_4.png)

### Структура диаграммы

- **Врач-травматолог (Doctor)**: Пользователь, который взаимодействует с системой для ввода данных пациента и просмотра результатов прогноза.
- **Веб-интерфейс (UI)**: Интерфейс пользователя, через который врач взаимодействует с системой. Обрабатывает ввод данных, отображает результаты и взаимодействует с бэкендом.
- **Backend API (Backend)**: Серверная часть приложения, обрабатывающая запросы от UI, взаимодействующая с базой данных и ML-моделью.
- **База данных PostgreSQL (DB)**: Хранит информацию о пациентах, включая их данные и историю операций.
- **ML Модель (Predict) (ML_Model)**: Модель машинного обучения, используемая для прогнозирования результата операции на основе данных пациента.

### Отношения и взаимодействия

- **Doctor -> UI**: Врач открывает главный модуль приложения.
- **UI -> Backend**: Веб-интерфейс отправляет запрос на получение ролевых прав от бэкенда.
- **Backend -> DB**: Бэкенд отправляет запрос в базу данных для проверки роли пользователя.
- **DB --> Backend**: База данных возвращает информацию о роли пользователя.
- **Backend --> UI**: Бэкенд отправляет ответ UI, подтверждающий роль пользователя.
- **Doctor -> UI**: Врач вводит данные пациента в интерфейсе.
- **UI -> Backend**: Веб-интерфейс отправляет POST-запрос с данными пациента на бэкенд.
- **Backend -> DB**: Бэкенд отправляет запрос в базу данных для получения информации о пациенте по его ID.
- **DB --> Backend**: База данных возвращает данные пациента.
- **Backend -> ML_Model**: Бэкенд отправляет запрос на выполнение прогноза от ML-модели.
- **activate ML_Model**: ML-модель активируется для выполнения прогноза.
- **ML_Model --> Backend**: ML-модель возвращает результат прогноза (вероятность и рекомендации).
- **deactivate ML_Model**: ML-модель деактивируется после выполнения прогноза.
- **Backend --> UI**: Бэкенд отправляет JSON-ответ с результатом прогноза.
- **UI --> Doctor**: Веб-интерфейс отображает результат прогноза врачу.
