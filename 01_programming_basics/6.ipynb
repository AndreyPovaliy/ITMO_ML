# Диаграмма модели предметной области и и словарь

![](./img/uml_2.png)

- **Patient (Пациент)**:
  - **Свойства**: Идентификатор, Дата рождения, Вес, Пол, Список сопутствующих заболеваний, Состояние суставов.
  - **Методы**: Расчет ИМТ (Индекса массы тела).

- **ClinicalData (Клинические данные)**:
  - **Свойства**: Возраст, Классификация артроза Келлина-Лауренс, ИМТ, Наличие диабета, Наличие остеопороза, Длительность операции.

- **Comorbidity (Сопутствующее заболевание)**:
  - **Свойства**: Тип заболевания, Уровень тяжести, Оценка риска ASA.

- **JointPathology (Патология сустава)**:
  - **Свойства**: Диагноз, Классификация Келлина-Лаура конкретного суства, Амплитуда движений, Оценка боли (в шкале ВАШ).

- **SurgeryLog (Журнал операций)**:
  - **Свойства**: Дата операции, Тип имплантации, Длительность операции, Используемая техника.

- **MLModelManager (Менеджер моделей машинного обучения)**:
  - **Свойства**: Набор признаков, Порог точности.
  - **Методы**: Обучение модели, Валидация модели, Прогнозирование исхода, Перезапуск модели.

- **MLModel (Модель машинного обучения)**:
  - **Свойства**: Версия модели, Точность, Чувствительность, Устойчивость, AUC-ROC.

- **PredictionResult (Результат прогноза)**:
  - **Свойства**: Вероятность успешного исхода, Критический риск.
  - **Методы**: Получение рекомендаций, Преобразование в текст.

- **BinaryClassifier (Бинарный классификатор)**:
  - **Методы**: Прогнозирование на основе клинических данных.

### Связи:
- **Patient и ClinicalData**: Отношение "многие к одному" (1..* - 1), что означает что у одного пациента может быть много клинических записей, но одна запись принадлежит одному пациенту.
- **Patient и JointPathology, Comorbidity, SurgeryLog**: Отношение "один к одному" (1 - 1), что означает что у одного пациента может быть только одна запись об состояниях сустава, сопутствующих заболеваниях и операциях.
- **ClinicalData и MLModelManager**: Отношение "один ко многим" (1 - *), что означает что на основе одной клинической записи может быть создано несколько моделей машинного обучения.
- **MLModel и BinaryClassifier**: Отношение "один ко многим" (1 - *), что означает что одна модель машинного обучения может использоваться многими бинарными классификаторами.
- **BinaryClassifier и PredictionResult**: Отношение "один ко многим" (1 - *), что означает что один классификатор может выдавать несколько результатов прогноза.
