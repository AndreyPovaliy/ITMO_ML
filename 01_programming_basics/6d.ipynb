# Типизация - типы данных

*реализация на языке Kotlin*

```kotlin
import java.util.Date
import java.time.LocalDate
import kotlin.math.round
```

```kotlin
class Comorbidity(val type: String, 
                        val severity: Int, 
                        var isContraindication: Boolean)
```

```kotlin
class JointPathology(val diagnosis: String, 
                            val kLStage: Int, 
                            val baselineROM: Float, 
                            val painScoreVAS: Int)
```

```kotlin
enum class ImplantType(val type: String) {
    HIP("Hip"),
    KNEE("Knee")
}
```

```kotlin
enum class Gender { MALE, FEMALE }
```

```kotlin
enum class Technique(val technique: String) {
    MINIMALLY_INVASIVE("Minimally Invasive"),
    OPEN_SURGERY("Open Surgery")
}
```

```kotlin
data class SurgeryLog(val operationDate: Date, 
                        val implant: ImplantType, 
                        val operationDuration: Double, 
                        val techniqueUsed: Technique)
```

```kotlin
data class Patient(
    val id: Int,
    val dateOfBirth: LocalDate,
    val weightKg: Float,
    val heightCm: Float,
    val gender: Gender,
    val comorbidities: List<Comorbidity>,
    val jointStatus: JointPathology
) {
    fun calculateBMI(): Float? {
        if (weightKg <= 0f || heightCm <= 0f) return null
        val heightM = heightCm / 100f
        return weightKg / (heightM * heightM)
    }
}
```

Особенность использовния `Kotlin`: особый тип данных в структуре языка которы позволяет возварщать функции или `Float` или `null`.

Это нужно при проведении проверкит деления на 0 в теле условния `(weightKg <= 0f || heightCm <= 0f)`. Таким образом избегаем деления на ноль, в kotlin ошибка`ArithmeticException`.

```kotlin
data class ClinicalData(
    val age: Double,
    val KellgrenLawrenceStage: Int,
    val BMI: Int,
    val diabetesPresent: Boolean,
    val osteoporosisPresent: Boolean,
    val surgeryDurationHours: Int
)
```
